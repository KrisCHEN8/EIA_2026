```python
import pandas as pd
import numpy as np
```

```python
years = range(2021, 2026)

# Load the daily dispatch data
df = pd.read_csv('./DH production mix/daily_priority_dispatch_2024.csv')

# Ensure the 'date' column is in datetime format
df['date'] = pd.to_datetime(df['date'])

priority_sources = ["Waste heat", "Electricity", "Biomass", "Fossil fuel"]

# Group by the month period and sum the daily total production (and category sources)
monthly_production = df.groupby(df['date'].dt.to_period('M'))[[
    'daily_total_mwh', 
    'Waste heat', 
    'Electricity', 
    'Biomass', 
    'Fossil fuel'
]].sum().reset_index()

# Display the monthly summary
print(monthly_production)
```
